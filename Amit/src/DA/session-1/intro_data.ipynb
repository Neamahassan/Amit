```python
def array_reshape(arr, rows , cols):
    try:
        arr = np.array(arr)
        return arr.reshape(rows, cols)
    except (TypeError, ValueError) as e:
        raise ValueError(f"Invalid reshape operatetion: {e}")
    except (TypeError, ValueError) as s:
        raise ValueError(f"Invalid reshape operatetion: {s}")
```

```python
def convert_dtype(arr,dtype):
    try:

        arr = np.array(arr)
        return arr.astype(dtype)
    except TypeError as e:
        raise TypeError(f"Invalid target data type :{dtype}") from e
```
