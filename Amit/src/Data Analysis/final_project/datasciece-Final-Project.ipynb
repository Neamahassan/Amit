###  To fix this issue You will fill these values with ........... for numeric and ......for categorical

```python
# trying to fill missing values in my datset using for loop 

# the station columns have only a few missing rows (197), so it's safe to just drop them

gbf = gbf.dropna(subset=['start_station_id','end_station_id','start_station_name', 'end_station_name'])

# now loop on the rest of the columns and fill the numeric ones with the median
for col in gbf.columns:
    if gbf[col].dtype in [' float64', 'int64']:
        gbf[col] = gbf[col].fillna(gbf[col].median())
```

```python
#check the Nulls 
gbf.isnull().sum()
```

# <a id="8"> 8) . What the average duration of trip for the categories of user type?</a>

When comparing the average trip duration across user categories, a clear behavioral difference emerges: casual Customers take significantly longer rides—averaging around 24 minutes—compared to Subscribers, whose average trip is roughly 11 minutes. This contrast highlights that regular members use the service with high efficiency for targeted daily commuting, whereas casual users or tourists tend to take more relaxed, extended trips.
