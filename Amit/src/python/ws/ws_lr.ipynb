```python
plt.figure(figsize=(3, 3))
sns.pairplot(
    data,
    x_vars=['YearsExperience'],
    y_vars=['Salary'],
    height=3,
    kind='scatter',
)
plt.xlabel('years')
plt.ylabel('salary')
plt.title('salary prediction')
plt.show()
```
