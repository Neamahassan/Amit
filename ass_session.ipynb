```python
import re

def decode_message(encoded: str) -> str:
    core = re.sub(r'[^A-Za-z\s]', '', encoded)
    core = core.strip() 
    print(f"Step 1 - Core extracted: '{core}'")

    words = core.split()
    first_word, second_word = words[0], words[1]

    decoded_first = first_word[::-1]
    print(f"Step 2 - Reversed first word: '{first_word}' -> '{decoded_first}'")

    vowels = "AEIOUaeiou"
    decoded_second = "".join(ch for ch in second_word if ch not in vowels)
    print(f"Step 3 - Vowels stripped from second word: '{second_word}' -> '{decoded_second}'")

    final_message = f"{decoded_first} {decoded_second}"
    print(f"Step 4 - Final decoded message: '{final_message}'")

    return final_message

encoded_message = "###! ! @mocleW EPGTQ!!!6789"
result = decode_message(encoded_message)
print("\nDecoded Message:", result)
```

```python
def process_email(email: str):
    if email.count("@") != 1:
        return "Invalid email"

    username, domain_part = email.split("@")

    if "." not in domain_part:
        return "Invalid email"

    print(f"Step 1 - Validation passed for: '{email}'")

    print(f"Step 2 - Username: '{username}'")

    last_dot_index = domain_part.rfind(".")
    domain = domain_part[:last_dot_index]
    print(f"Step 3 - Domain: '{domain}'")

    if email.endswith(".com"):
        ending_result = "Commercial Domain"
    elif email.endswith(".edu"):
        ending_result = "Educational Domain"
    else:
        ending_result = "Other Domain"
    print(f"Step 4 - Domain Ending Check: '{ending_result}'")

    return {
        "Username": username,
        "Domain": domain,
        "Domain Type": ending_result
    }

email_address = "Amit_ml@gmail.edu"
result = process_email(email_address)
print("\nFinal Result:", result)
```
