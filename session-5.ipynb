```python
from abc import ABC, abstractmethod
   
class Animal(ABC):
    @abstractmethod
    def move(self):
        pass

class Bird(Animal):
    def move(self):
        print("move from Bird")

class Cat(Animal):
    def move(self):
        print("move from Cat")
```

```python
class Student:
    def __init__(self, name, age) -> None:
        self.__name = name
        self.age = age

    def std_info(self):
        return f"my name is {self.__name} and my age is {self.age}"
```

```python
s1 = Student("neama", 24)

print(s1.std_info())
```

```python
class Student:
    def __init__(self, age) -> None:
        self._age = age

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("age must be positive")
        self._age = value
```
